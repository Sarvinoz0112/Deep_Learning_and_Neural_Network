## 2. What is a tensor?

A **tensor** is a container of numbers. It is the basic building block of deep learning: images, text, audio and tables are all turned into tensors before a model can use them.

For example, a colour image can be stored as a tensor of shape `[3, 224, 224]`, meaning `[colour_channels, height, width]`.

Tensors come in different numbers of dimensions:

| Name | What is it? | Number of dimensions (`ndim`) |
| ----- | ----- | ----- |
| **scalar** | a single number | 0 |
| **vector** | a list of numbers | 1 |
| **matrix** | a table of numbers (rows and columns) | 2 |
| **tensor** | numbers arranged in any number of dimensions | 0, 1, 2, 3, ... |

![scalar vector matrix tensor and what they look like](https://raw.githubusercontent.com/mrdbourke/pytorch-deep-learning/main/images/00-scalar-vector-matrix-tensor.png)

Two attributes describe every tensor:
* `ndim` - how many dimensions it has
* `shape` - how many elements there are along each dimension

```python
# Scalar: a single number, 0 dimensions
scalar = torch.tensor(7)
print(scalar)
print("ndim: ", scalar.ndim)
print("shape:", scalar.shape)
print("as a Python number:", scalar.item())   # .item() only works on one-element tensors
```

```python
# Vector: 1 dimension
vector = torch.tensor([7, 7])
print(vector)
print("ndim: ", vector.ndim)
print("shape:", vector.shape)
```

```python
# Matrix: 2 dimensions (rows, columns)
MATRIX = torch.tensor([[7, 8],
                       [9, 10]])
print(MATRIX)
print("ndim: ", MATRIX.ndim)
print("shape:", MATRIX.shape)
```

```python
# Tensor: 3 dimensions here
TENSOR = torch.tensor([[[1, 2, 3],
                        [3, 6, 9],
                        [2, 4, 5]]])
print(TENSOR)
print("ndim: ", TENSOR.ndim)
print("shape:", TENSOR.shape)   # 1 block of 3 rows and 3 columns
```

**Tip:** the number of dimensions equals the number of opening square brackets `[` at the start of the tensor. `[[[` means 3 dimensions.

The shape is read from the outside in: `torch.Size([1, 3, 3])` means 1 block, each with 3 rows, each with 3 values.

## 3. Creating tensors

You will rarely type tensors by hand. A neural network starts with tensors full of **random numbers** and then adjusts them as it learns:

`start with random numbers -> look at data -> update the numbers -> look at data -> update the numbers ...`

The most useful functions for creating tensors are:

| Function | What it creates |
| ----- | ----- |
| `torch.rand(size)` | random numbers between 0 and 1 |
| `torch.zeros(size)` | all zeros |
| `torch.ones(size)` | all ones |
| `torch.arange(start, end, step)` | a range of numbers (`end` is not included) |
| `torch.zeros_like(x)`, `torch.ones_like(x)` | zeros or ones with the same shape as `x` |

```python
# Random tensor with 3 rows and 4 columns
random_tensor = torch.rand(size=(3, 4))
print(random_tensor)

# Random tensor in the shape of an image: [colour_channels, height, width]
random_image = torch.rand(size=(3, 224, 224))
print(random_image.shape, random_image.ndim)
```

```python
# Zeros and ones
zeros = torch.zeros(size=(3, 4))
ones = torch.ones(size=(3, 4))
print(zeros)
print(ones)
```

```python
# A range of numbers: start=0, end=10 (not included), step=1
zero_to_nine = torch.arange(start=0, end=10, step=1)
print(zero_to_nine)

# Zeros with the same shape as another tensor
print(torch.zeros_like(zero_to_nine))
```

## 4. Tensor datatypes

Every tensor has a datatype (`dtype`). The ones you will meet most often:

| dtype | Meaning | When you see it |
| ----- | ----- | ----- |
| `torch.float32` | decimal numbers, 32 bits | the default for decimal values, used by most models |
| `torch.int64` | whole numbers, 64 bits | the default for whole numbers, used for labels and indexes |
| `torch.float16` | decimal numbers, 16 bits | smaller and faster, but less precise |

You choose the datatype with the `dtype` parameter, and you change it later with `.type()`.

```python
# Decimal values give float32 by default
float_32_tensor = torch.tensor([3.0, 6.0, 9.0])
print(float_32_tensor.dtype)

# Whole numbers give int64 by default
int_tensor = torch.tensor([3, 6, 9])
print(int_tensor.dtype)

# Choose the datatype when creating the tensor
float_16_tensor = torch.tensor([3.0, 6.0, 9.0], dtype=torch.float16)
print(float_16_tensor.dtype)

# Change the datatype of an existing tensor
converted = int_tensor.type(torch.float32)
print(converted, converted.dtype)
```

## 6. Tensor operations

A neural network is built from a small set of operations on tensors: addition, subtraction, multiplication, division and **matrix multiplication**.

### 6.1 Basic operations

These work element by element.

```python
tensor = torch.tensor([1, 2, 3])

print(tensor + 10)   # add 10 to every element
print(tensor - 10)   # subtract 10 from every element
print(tensor * 10)   # multiply every element by 10
print(tensor / 10)   # divide every element by 10

# The original tensor does not change unless you reassign it
print(tensor)
tensor = tensor + 10
print(tensor)
```

### 6.2 Element-wise multiplication vs matrix multiplication

For a tensor with values `[1, 2, 3]`:

| Operation | Calculation | Code |
| ----- | ----- | ----- |
| **Element-wise multiplication** | `[1*1, 2*2, 3*3]` = `[1, 4, 9]` | `tensor * tensor` |
| **Matrix multiplication** | `1*1 + 2*2 + 3*3` = `14` | `torch.matmul(tensor, tensor)` or `tensor @ tensor` |

```python
tensor = torch.tensor([1, 2, 3])

print("Element-wise:", tensor * tensor)
print("Matrix multiplication:", torch.matmul(tensor, tensor))
print("Matrix multiplication with @:", tensor @ tensor)
```

### 6.3 The two rules of matrix multiplication

1. The **inner dimensions** must match:
  * `(3, 2) @ (3, 2)` will not work
  * `(2, 3) @ (3, 2)` will work
  * `(3, 2) @ (2, 3)` will work
2. The result has the shape of the **outer dimensions**:
  * `(2, 3) @ (3, 2)` -> `(2, 2)`
  * `(3, 2) @ (2, 3)` -> `(3, 3)`

Breaking rule 1 gives a **shape error**, one of the most common errors in deep learning.

```python
tensor_A = torch.tensor([[1., 2.],
                         [3., 4.],
                         [5., 6.]])     # shape (3, 2)

tensor_B = torch.tensor([[7., 10.],
                         [8., 11.],
                         [9., 12.]])    # shape (3, 2)

# (3, 2) @ (3, 2): the inner dimensions 2 and 3 do not match, so this fails
try:
    torch.matmul(tensor_A, tensor_B)
except RuntimeError as error:
    print("Error:", error)
```

To fix it, **transpose** one of the tensors with `.T`. Transposing swaps rows and columns, so a `(3, 2)` tensor becomes `(2, 3)`.

```python
print("tensor_B shape:  ", tensor_B.shape)
print("tensor_B.T shape:", tensor_B.T.shape)

# (3, 2) @ (2, 3): the inner dimensions match, the result is (3, 3)
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print("Output shape:", output.shape)
```

## 7. Aggregation (min, max, mean, sum)

Aggregation turns many values into fewer values.

`argmin()` and `argmax()` return the **position (index)** of the smallest and largest value, not the value itself.

```python
x = torch.arange(0, 100, 10)
print(x)

print(f"Minimum: {x.min()}")
print(f"Maximum: {x.max()}")
print(f"Sum: {x.sum()}")

# mean() needs a float tensor, so convert the datatype first
print(f"Mean: {x.type(torch.float32).mean()}")

print(f"Index of the minimum value: {x.argmin()}")
print(f"Index of the maximum value: {x.argmax()}")
```

## 8. Changing the shape of a tensor

Models need tensors in the right shape. These methods change the shape without changing the values:

| Method | What it does |
| ----- | ----- |
| `x.reshape(shape)` | gives `x` a new shape (the total number of elements must stay the same) |
| `torch.stack([a, b], dim)` | joins tensors along a **new** dimension |
| `x.squeeze()` | removes all dimensions of size 1 |
| `x.unsqueeze(dim)` | adds a dimension of size 1 at position `dim` |
| `x.permute(dims)` | rearranges the order of the dimensions |

```python
x = torch.arange(1., 9.)
print(x, x.shape)

# reshape: 8 elements can become 2 rows of 4
x_reshaped = x.reshape(2, 4)
print(x_reshaped, x_reshaped.shape)
```

```python
# stack: join tensors along a new dimension
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])

print(torch.stack([a, b], dim=0))   # a and b become rows,    shape (2, 3)
print(torch.stack([a, b], dim=1))   # a and b become columns, shape (3, 2)
```

```python
# squeeze: remove dimensions of size 1
y = torch.rand(1, 5)
y_squeezed = y.squeeze()
print("Before squeeze:", y.shape)
print("After squeeze: ", y_squeezed.shape)

# unsqueeze: add a dimension of size 1
y_unsqueezed = y_squeezed.unsqueeze(dim=0)
print("After unsqueeze(dim=0):", y_unsqueezed.shape)
```

```python
# permute: rearrange the dimensions
# Example: an image stored as [height, width, colour_channels]
image = torch.rand(size=(224, 224, 3))

# PyTorch models expect [colour_channels, height, width]
image_permuted = image.permute(2, 0, 1)   # old dimension 2 goes first, then 0, then 1

print("Before permute:", image.shape)
print("After permute: ", image_permuted.shape)
```

## 9. Indexing (selecting data from tensors)

Indexing works like Python lists and NumPy arrays. You index from the outer dimension to the inner dimension. Use `:` to mean "everything in this dimension".

`torch.Size([1, 3, 3])`.

The dimensions go outer to inner.

That means there's 1 dimension of 3 by 3.

![example of different tensor dimensions](https://raw.githubusercontent.com/mrdbourke/pytorch-deep-learning/main/images/00-pytorch-different-tensor-dimensions.png)

> **Note:** You might've noticed me using lowercase letters for `scalar` and `vector` and uppercase letters for `MATRIX` and `TENSOR`. This was on purpose. In practice, you'll often see scalars and vectors denoted as lowercase letters such as `y` or `a`. And matrices and tensors denoted as uppercase letters such as `X` or `W`.
>
> You also might notice the names martrix and tensor used interchangably. This is common. Since in PyTorch you're often dealing with `torch.Tensor`s (hence the tensor name), however, the shape and dimensions of what's inside will dictate what it actually is.


```python
x = torch.arange(1, 10).reshape(1, 3, 3)
print(x, x.shape)
```

```python
print(x[0])          # the first (and only) block: a 3x3 matrix
print(x[0][0])       # the first row of that block
print(x[0][0][0])    # the first value of that row

# The same thing with commas
print(x[0, 1, 1])    # block 0, row 1, column 1 -> 5
print(x[0, 1, :])    # block 0, row 1, all columns -> [4, 5, 6]
print(x[0, :, 1])    # block 0, all rows, column 1 -> [2, 5, 8]
```

## 10. PyTorch and NumPy

Data often arrives as a NumPy array, and plotting libraries expect NumPy arrays, so you need to convert in both directions:

* `torch.from_numpy(array)` - NumPy array -> PyTorch tensor
* `tensor.numpy()` - PyTorch tensor -> NumPy array

Note: NumPy uses `float64` by default while PyTorch uses `float32`. The conversion keeps the original datatype.

```python
import numpy as np

# NumPy array -> tensor
array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
print(array, array.dtype)
print(tensor)

# Convert to float32 if you need the PyTorch default
print(tensor.type(torch.float32).dtype)

# Tensor -> NumPy array
tensor = torch.ones(7)
numpy_array = tensor.numpy()
print(numpy_array, numpy_array.dtype)
```

## 11. Reproducibility (random seed)

`torch.rand()` gives different numbers every time. That makes experiments hard to repeat.

`torch.manual_seed(seed)` fixes the starting point of the random number generator, so the same seed always gives the same "random" numbers.

Important: call `torch.manual_seed()` again before **each** random call that you want to repeat.

```python
# Without a seed: two different tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)
print(random_tensor_A == random_tensor_B)
```

```python
# With a seed: two identical tensors
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)    # set the seed again before the second call
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```

## 12. Running tensors on a GPU

A GPU does the maths of neural networks (mainly matrix multiplication) much faster than a CPU.

The standard pattern is **device-agnostic code**: use the GPU if there is one, otherwise use the CPU. The same code then runs on any computer.

```python
# Is a GPU available?
print(torch.cuda.is_available())

# Device-agnostic setup
device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

Move a tensor to a device with `.to(device)`. This returns a copy, so reassign it if you want to keep the result.

Two tensors must be on the **same device** before you can combine them.

```python
tensor = torch.tensor([1, 2, 3])
print(tensor, tensor.device)

tensor_on_device = tensor.to(device)
print(tensor_on_device, tensor_on_device.device)
```

NumPy only works on the CPU. Calling `.numpy()` on a GPU tensor gives an error, so bring the tensor back with `.cpu()` first.

```python
tensor_back_on_cpu = tensor_on_device.cpu().numpy()
print(tensor_back_on_cpu)
```
